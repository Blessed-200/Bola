```python
import pandas as pd
from datetime import timedelta

# -------------------------------------------------------------------
# 1. Carga robusta del CSV
# -------------------------------------------------------------------
file_path = "US_Lottery_Draws_Master_modified.csv"

# Leer el archivo; si hay valores faltantes se conservan como NaN
df = pd.read_csv(file_path, dtype=str)  # todo como string para evitar problemas

# Normalizar nombres de columnas (por si hay espacios)
df.columns = df.columns.str.strip()

# Columnas que nos interesan
required_cols = ['Estado', 'Sesión', 'Fecha', 'Fijo', 'Corrido_1', 'Corrido_2']
# Eliminar filas donde alguna de estas columnas sea NaN (o cadena vacía)
df = df.dropna(subset=required_cols)
df = df[df[required_cols].ne('').all(axis=1)]

# Convertir Fecha a datetime
df['Fecha'] = pd.to_datetime(df['Fecha'], errors='coerce')
df = df.dropna(subset=['Fecha'])  # quitar fechas no válidas

# Asegurar que los números tengan dos dígitos (relleno con ceros)
for col in ['Fijo', 'Corrido_1', 'Corrido_2']:
    # Convertir a string, quitar espacios, rellenar a 2 caracteres con '0' a la izquierda
    df[col] = df[col].astype(str).str.strip().str.zfill(2)

# -------------------------------------------------------------------
# 2. Filtrar sorteos GA Midday (nuestros objetivos)
# -------------------------------------------------------------------
ga_midday = df[(df['Estado'] == 'GA') & (df['Sesión'] == 'Midday')].copy()
ga_midday = ga_midday.sort_values('Fecha').reset_index(drop=True)

# Nos interesa a partir del segundo día (porque necesitamos un día anterior)
if len(ga_midday) < 2:
    raise ValueError("No hay suficientes sorteos GA Midday para analizar.")

# Lista para almacenar los registros de coincidencias
origins = []

# -------------------------------------------------------------------
# 3. Para cada GA Midday (excepto el primero) buscar orígenes en el día anterior
# -------------------------------------------------------------------
for i in range(1, len(ga_midday)):
    target_row = ga_midday.iloc[i]
    target_date = target_row['Fecha']
    target_fijo = target_row['Fijo']          # string de 2 dígitos
    target_tens = target_fijo[0]              # decena
    target_units = target_fijo[1]             # unidad

    # Día anterior
    prev_date = target_date - timedelta(days=1)
    # Todos los sorteos de cualquier estado/sesión del día anterior
    prev_draws = df[df['Fecha'] == prev_date]

    if prev_draws.empty:
        # Si no hay datos del día anterior, simplemente continuamos
        continue

    # Recorrer cada sorteo del día anterior
    for _, src in prev_draws.iterrows():
        src_date = src['Fecha'].strftime('%Y-%m-%d')
        src_estado = src['Estado']
        src_sesion = src['Sesión']

        # Para cada uno de los tres números: Fijo, Corrido_1, Corrido_2
        for num_type in ['Fijo', 'Corrido_1', 'Corrido_2']:
            src_number = src[num_type]
            # Verificar que sea un string de 2 dígitos (debería serlo tras el zfill)
            if len(src_number) != 2:
                continue
            src_tens = src_number[0]
            src_units = src_number[1]

            # Comparar con la decena del Fijo objetivo
            if src_tens == target_tens:
                origins.append({
                    'target_date': target_date.strftime('%Y-%m-%d'),
                    'target_fijo': target_fijo,
                    'target_digit_type': 'decena',
                    'target_digit': target_tens,
                    'source_date': src_date,
                    'source_estado': src_estado,
                    'source_sesion': src_sesion,
                    'source_number_type': num_type,
                    'source_digit_type': 'decena'
                })
            if src_units == target_tens:
                origins.append({
                    'target_date': target_date.strftime('%Y-%m-%d'),
                    'target_fijo': target_fijo,
                    'target_digit_type': 'decena',
                    'target_digit': target_tens,
                    'source_date': src_date,
                    'source_estado': src_estado,
                    'source_sesion': src_sesion,
                    'source_number_type': num_type,
                    'source_digit_type': 'unidad'
                })

            # Comparar con la unidad del Fijo objetivo
            if src_tens == target_units:
                origins.append({
                    'target_date': target_date.strftime('%Y-%m-%d'),
                    'target_fijo': target_fijo,
                    'target_digit_type': 'unidad',
                    'target_digit': target_units,
                    'source_date': src_date,
                    'source_estado': src_estado,
                    'source_sesion': src_sesion,
                    'source_number_type': num_type,
                    'source_digit_type': 'decena'
                })
            if src_units == target_units:
                origins.append({
                    'target_date': target_date.strftime('%Y-%m-%d'),
                    'target_fijo': target_fijo,
                    'target_digit_type': 'unidad',
                    'target_digit': target_units,
                    'source_date': src_date,
                    'source_estado': src_estado,
                    'source_sesion': src_sesion,
                    'source_number_type': num_type,
                    'source_digit_type': 'unidad'
                })

# -------------------------------------------------------------------
# 4. Guardar resultados
# -------------------------------------------------------------------
origins_df = pd.DataFrame(origins)
output_file = "ga_midday_fijo_origins.csv"
origins_df.to_csv(output_file, index=False, encoding='utf-8-sig')
print(f"Archivo guardado: {output_file}")
print(f"Total de coincidencias registradas: {len(origins_df)}")
```

```python
import pandas as pd
from collections import defaultdict
import sys

# -------------------------------------------------------------------
# 1. Cargar archivo de orígenes
# -------------------------------------------------------------------
origins = pd.read_csv("ga_midday_fijo_origins.csv")
print(f"Total de coincidencias cargadas: {len(origins)}")

# Días únicos de GA Midday
dias_target = sorted(origins['target_date'].unique())
print(f"Días únicos (GA Midday) en el archivo: {len(dias_target)}")

# -------------------------------------------------------------------
# 2. Preparar datos para Set Cover
# -------------------------------------------------------------------
# Para cada tipo de dígito (decena / unidad) crearemos:
# - Una lista de reglas (tuplas: estado, sesion, num_type, digit_type)
# - Un diccionario día -> conjunto de índices de reglas que aciertan

def preparar_cobertura(target_digit_type):
    df_tipo = origins[origins['target_digit_type'] == target_digit_type].copy()
    # Crear un identificador de regla a partir de las columnas fuente
    df_tipo['regla'] = list(zip(
        df_tipo['source_estado'],
        df_tipo['source_sesion'],
        df_tipo['source_number_type'],
        df_tipo['source_digit_type']
    ))
    # Lista de reglas únicas
    reglas_unicas = sorted(df_tipo['regla'].unique())
    regla_a_idx = {r: i for i, r in enumerate(reglas_unicas)}
    # Cobertura: día -> set de índices de reglas que aciertan
    cobertura = defaultdict(set)
    for _, row in df_tipo.iterrows():
        dia = row['target_date']
        idx = regla_a_idx[row['regla']]
        cobertura[dia].add(idx)
    return reglas_unicas, cobertura

reglas_decena, cobertura_decena = preparar_cobertura('decena')
reglas_unidad, cobertura_unidad = preparar_cobertura('unidad')

print(f"Reglas candidatas para DECENA: {len(reglas_decena)}")
print(f"Reglas candidatas para UNIDAD: {len(reglas_unidad)}")

# -------------------------------------------------------------------
# 3. Función para resolver Set Cover (mínimo número de reglas)
# -------------------------------------------------------------------
def set_cover_greedy(universo, cobertura, reglas):
    """
    universo: lista de todos los días
    cobertura: dict día -> set de índices de reglas que lo cubren
    reglas: lista de reglas
    Retorna lista de índices seleccionados
    """
    # Invertir cobertura: regla -> set de días que cubre
    dias_por_regla = defaultdict(set)
    for dia, reglas_idx in cobertura.items():
        for idx in reglas_idx:
            dias_por_regla[idx].add(dia)

    universo_set = set(universo)
    cubiertos = set()
    seleccionadas = []

    while cubiertos != universo_set:
        # Elegir la regla que cubre más días aún no cubiertos
        mejor_idx = None
        mejor_cubierta = -1
        for idx, dias in dias_por_regla.items():
            if idx in seleccionadas:
                continue
            nuevos = len(dias - cubiertos)
            if nuevos > mejor_cubierta:
                mejor_cubierta = nuevos
                mejor_idx = idx
        if mejor_idx is None or mejor_cubierta == 0:
            break
        seleccionadas.append(mejor_idx)
        cubiertos.update(dias_por_regla[mejor_idx])
    return seleccionadas

def set_cover_ilp(universo, cobertura, reglas):
    """Resuelve Set Cover exacto usando PuLP (si está disponible)."""
    try:
        from pulp import LpProblem, LpMinimize, LpVariable, lpSum, LpBinary, value
    except ImportError:
        print("PuLP no está instalado. Usando método voraz (greedy).")
        return set_cover_greedy(universo, cobertura, reglas)

    dias_idx = {dia: i for i, dia in enumerate(universo)}
    n_dias = len(universo)
    n_reglas = len(reglas)

    # Matriz de cobertura: cobertura_mat[i][j] = 1 si la regla j cubre el día i
    cobertura_mat = [[0]*n_reglas for _ in range(n_dias)]
    for dia, reglas_idx in cobertura.items():
        i = dias_idx[dia]
        for j in reglas_idx:
            cobertura_mat[i][j] = 1

    prob = LpProblem("SetCover", LpMinimize)
    x = [LpVariable(f"r_{j}", cat=LpBinary) for j in range(n_reglas)]
    prob += lpSum(x)
    for i in range(n_dias):
        prob += lpSum(cobertura_mat[i][j] * x[j] for j in range(n_reglas)) >= 1

    prob.solve()
    if prob.status != 1:  # no óptimo
        return set_cover_greedy(universo, cobertura, reglas)
    seleccionadas = [j for j in range(n_reglas) if value(x[j]) == 1]
    return seleccionadas

# Elegir método: PuLP si disponible, sino greedy
try:
    import pulp
    usar_ilp = True
except ImportError:
    usar_ilp = False
    print("PuLP no encontrado. Usando algoritmo voraz (greedy). Instala con: pip install pulp")

# -------------------------------------------------------------------
# 4. Resolver cobertura para decena y para unidad
# -------------------------------------------------------------------
universo_decena = list(cobertura_decena.keys())
universo_unidad = list(cobertura_unidad.keys())

if usar_ilp:
    seleccion_decena = set_cover_ilp(universo_decena, cobertura_decena, reglas_decena)
    seleccion_unidad = set_cover_ilp(universo_unidad, cobertura_unidad, reglas_unidad)
else:
    seleccion_decena = set_cover_greedy(universo_decena, cobertura_decena, reglas_decena)
    seleccion_unidad = set_cover_greedy(universo_unidad, cobertura_unidad, reglas_unidad)

# -------------------------------------------------------------------
# 5. Mostrar resultados
# -------------------------------------------------------------------
print("\n" + "="*70)
print("REGLAS MÍNIMAS PARA LA DECENA")
print("="*70)
for idx in seleccion_decena:
    estado, sesion, num_type, digit_type = reglas_decena[idx]
    print(f"- Tomar dígito {digit_type} del {num_type} de {estado} {sesion} del día anterior")

print("\n" + "="*70)
print("REGLAS MÍNIMAS PARA LA UNIDAD")
print("="*70)
for idx in seleccion_unidad:
    estado, sesion, num_type, digit_type = reglas_unidad[idx]
    print(f"- Tomar dígito {digit_type} del {num_type} de {estado} {sesion} del día anterior")

print(f"\nTotal reglas: {len(seleccion_decena)} decena + {len(seleccion_unidad)} unidad = {len(seleccion_decena)+len(seleccion_unidad)}")

# -------------------------------------------------------------------
# 6. Verificar cobertura 100% en histórico
# -------------------------------------------------------------------
def verificar_cobertura(universo, cobertura, seleccionadas):
    cubiertos = set()
    for dia in universo:
        if any(idx in cobertura[dia] for idx in seleccionadas):
            cubiertos.add(dia)
    return len(cubiertos) == len(universo), len(cubiertos), len(universo)

ok_dec, n_dias_dec, total_dec = verificar_cobertura(universo_decena, cobertura_decena, seleccion_decena)
ok_uni, n_dias_uni, total_uni = verificar_cobertura(universo_unidad, cobertura_unidad, seleccion_unidad)

print(f"\nCobertura decena: {n_dias_dec}/{total_dec} días ({100*n_dias_dec/total_dec:.2f}%)")
print(f"Cobertura unidad: {n_dias_uni}/{total_uni} días ({100*n_dias_uni/total_uni:.2f}%)")
if ok_dec and ok_uni:
    print("✅ ¡Cobertura perfecta! Todas las decenas y unidades de todos los días quedan explicadas.")
else:
    print("⚠️ Hay días sin cubrir. Revisar datos o añadir reglas adicionales.")

# -------------------------------------------------------------------
# 7. Cuántos números generan estas reglas (media histórica)
# -------------------------------------------------------------------
# Para cada día, obtenemos los dígitos predichos por las reglas seleccionadas
# y contamos el producto cartesiano.

def digitos_predichos(dia, cobertura, reglas, seleccionadas):
    digitos = set()
    for idx in seleccionadas:
        if idx in cobertura.get(dia, set()):
            # El dígito que daría esta regla sería el mismo que la regla "detecta"
            # Para saber qué dígito predice, necesitamos buscar en el día anterior
            # la fuente indicada por la regla. Pero como aquí solo validamos cobertura,
            # podemos tomar el target_digit de esa coincidencia en el archivo origins.
            # Sin embargo, para este análisis de cobertura no es necesario; solo
            # nos interesa la cantidad de dígitos diferentes que resultan.
            pass
    # En su lugar, podemos calcular para cada día el conjunto de dígitos de decena/unidad
    # que realmente saldrían de aplicar las reglas a los datos del día anterior.
    # Esto requiere cargar el csv original para obtener los valores fuente.
    # Lo haremos de forma práctica a continuación.

# Mejor: cargar datos originales y aplicar las reglas
df_orig = pd.read_csv("US_Lottery_Draws_Master_modified.csv", dtype=str)
df_orig.columns = df_orig.columns.str.strip()
for col in ['Fijo', 'Corrido_1', 'Corrido_2']:
    df_orig[col] = df_orig[col].str.strip().str.zfill(2)
df_orig['Fecha'] = pd.to_datetime(df_orig['Fecha'], errors='coerce')
df_orig = df_orig.dropna(subset=['Fecha', 'Fijo', 'Corrido_1', 'Corrido_2'])

# Construir lookup rápido del día anterior
from datetime import timedelta

# Para cada día en el histórico, aplicamos las reglas y vemos los dígitos generados
combinaciones_por_dia = []
for dia in dias_target:
    fecha_obj = pd.to_datetime(dia)
    prev = fecha_obj - timedelta(days=1)
    prev_draws = df_orig[df_orig['Fecha'] == prev]
    if prev_draws.empty:
        continue
    # Obtener decenas candidatas
    decenas = set()
    for idx in seleccion_decena:
        estado, sesion, num_type, digit_type = reglas_decena[idx]
        # Filtrar el sorteo que coincide en estado y sesión
        match = prev_draws[(prev_draws['Estado'] == estado) & (prev_draws['Sesión'] == sesion)]
        if not match.empty:
            num = match.iloc[0][num_type]
            if digit_type == 'decena':
                dig = num[0]
            else:
                dig = num[1]
            decenas.add(dig)
    unidades = set()
    for idx in seleccion_unidad:
        estado, sesion, num_type, digit_type = reglas_unidad[idx]
        match = prev_draws[(prev_draws['Estado'] == estado) & (prev_draws['Sesión'] == sesion)]
        if not match.empty:
            num = match.iloc[0][num_type]
            if digit_type == 'decena':
                dig = num[0]
            else:
                dig = num[1]
            unidades.add(dig)
    # Producto cartesiano
    combis = len(decenas) * len(unidades)
    combinaciones_por_dia.append(combis)

if combinaciones_por_dia:
    import numpy as np
    media_combinaciones = np.mean(combinaciones_por_dia)
    max_combinaciones = max(combinaciones_por_dia)
    print(f"\nPromedio de números generados por día: {media_combinaciones:.1f}")
    print(f"Máximo generado en un día: {max_combinaciones}")
```

```python
import pandas as pd

# 1. Cargar orígenes
origins = pd.read_csv('ga_midday_fijo_origins.csv')
total_dias = origins['target_date'].nunique()

# 2. Calcular la frecuencia de éxito de cada regla individual
def analizar_eficiencia(tipo):
    df_tipo = origins[origins['target_digit_type'] == tipo].copy()
    df_tipo['regla'] = df_tipo['source_estado'] + ' ' + df_tipo['source_sesion'] + ' (' + df_tipo['source_number_type'] + '_' + df_tipo['source_digit_type'] + ')'

    # Contar en cuántos días únicos acierta cada regla
    rank = df_tipo.groupby('regla')['target_date'].nunique().reset_index()
    rank.columns = ['Regla', 'Dias_Acertados']
    rank['Probabilidad_%'] = (rank['Dias_Acertados'] / total_dias) * 100
    return rank.sort_values(by='Probabilidad_%', ascending=False)

rank_decena = analizar_eficiencia('decena')
rank_unidad = analizar_eficiencia('unidad')

print(f"--- TOP 10 REGLAS PARA DECENA (Total días analizados: {total_dias}) ---")
display(rank_decena.head(10))

print(f"\n--- TOP 10 REGLAS PARA UNIDAD ---")
display(rank_unidad.head(10))
```

### Simulación con las mejores reglas
Ahora, en lugar de usar 80 reglas, probaremos qué pasa si usamos solo las 3 mejores para la decena y las 3 mejores para la unidad. Esto reducirá drásticamente la cantidad de números generados (máximo 9 números).

```python
def simular_top_n(n_reglas):
    top_dec = rank_decena.head(n_reglas)['Regla'].tolist()
    top_uni = rank_unidad.head(n_reglas)['Regla'].tolist()

    # Filtrar orígenes solo para estas reglas
    df_dec = origins[(origins['target_digit_type'] == 'decena') &
                     ((origins['source_estado'] + ' ' + origins['source_sesion'] + ' (' + origins['source_number_type'] + '_' + origins['source_digit_type'] + ')').isin(top_dec))]

    df_uni = origins[(origins['target_digit_type'] == 'unidad') &
                     ((origins['source_estado'] + ' ' + origins['source_sesion'] + ' (' + origins['source_number_type'] + '_' + origins['source_digit_type'] + ')').isin(top_uni))]

    # Dias donde AMBAS (decena y unidad) fueron capturadas por el Top N
    dias_exito = set(df_dec['target_date']) & set(df_uni['target_date'])

    print(f"Simulación usando Top {n_reglas} reglas:")
    print(f"- Números generados por día: máximo {n_reglas * n_reglas}")
    print(f"- Días con acierto garantizado: {len(dias_exito)} de {total_dias} ({len(dias_exito)/total_dias*100:.2f}%)")

simular_top_n(3)
simular_top_n(5)
```

```python
# 1. Analizar combinaciones específicas de reglas (Parejas de Éxito)
# Buscamos qué regla de Decena + qué regla de Unidad suelen acertar JUNTAS

def analizar_parejas_exito(n_top=15):
    # Preparar datos de aciertos de decenas
    df_dec = origins[origins['target_digit_type'] == 'decena'].copy()
    df_dec['regla_dec'] = df_dec['source_estado'] + ' ' + df_dec['source_sesion'] + ' (' + df_dec['source_number_type'] + '_' + df_dec['source_digit_type'] + ')'

    # Preparar datos de aciertos de unidades
    df_uni = origins[origins['target_digit_type'] == 'unidad'].copy()
    df_uni['regla_uni'] = df_uni['source_estado'] + ' ' + df_uni['source_sesion'] + ' (' + df_uni['source_number_type'] + '_' + df_uni['source_digit_type'] + ')'

    # Cruzar por fecha para ver qué pares de reglas acertaron el mismo día
    merged = pd.merge(df_dec[['target_date', 'regla_dec']], df_uni[['target_date', 'regla_uni']], on='target_date')

    # Contar frecuencia de la pareja de reglas (Estandarizado a 'Frecuencia')
    parejas = merged.groupby(['regla_dec', 'regla_uni']).size().reset_index(name='Frecuencia')
    parejas = parejas.sort_values(by='Frecuencia', ascending=False)

    print(f"--- TOP {n_top} PAREJAS DE REGLAS CON MAYOR SINERGIA ---")
    display(parejas.head(n_top))
    return parejas

parejas_top = analizar_parejas_exito(15)
```

```python
import pandas as pd
from datetime import timedelta

def generar_sugerencias_sinergicas(n_parejas=15):
    # 1. Obtener la última fecha en el dataset
    ultima_fecha = df_orig['Fecha'].max()
    datos_ayer = df_orig[df_orig['Fecha'] == ultima_fecha]

    print(f"Generando sugerencias basadas en datos del: {ultima_fecha.strftime('%Y-%m-%d')}")
    print(f"Para el sorteo GA Midday del día siguiente.\n")

    sugerencias = set()
    detalles = []

    # 2. Iterar sobre las top parejas sinérgicas
    for _, row in parejas_top.head(n_parejas).iterrows():
        r_dec = row['regla_dec']
        r_uni = row['regla_uni']

        def obtener_digito(regla_str, datos_fuente):
            # Extraer componentes del string de la regla
            partes = regla_str.split(' ')
            est = partes[0]
            ses = partes[1]
            # Extraer contenido de los paréntesis: 'Corrido_1_decena'
            contenido_par = regla_str.split('(')[1].split(')')[0]
            # Separar el tipo de número y el tipo de dígito
            # El tipo de dígito es siempre la última palabra tras el último guion bajo
            subpartes = contenido_par.split('_')
            dig_t = subpartes[-1] # 'decena' o 'unidad'
            num_t = '_'.join(subpartes[:-1]) # 'Corrido_1', 'Fijo', etc.

            match = datos_fuente[(datos_fuente['Estado'] == est) & (datos_fuente['Sesión'] == ses)]
            if not match.empty:
                val = match.iloc[0][num_t]
                return val[0] if dig_t == 'decena' else val[1]
            return None

        d = obtener_digito(r_dec, datos_ayer)
        u = obtener_digito(r_uni, datos_ayer)

        if d is not None and u is not None:
            numero = d + u
            sugerencias.add(numero)
            detalles.append({'Numero': numero, 'Sinergia_Hist': row['Frecuencia_Exito_Conjunto'], 'Regla_Dec': r_dec, 'Regla_Uni': u})

    # 3. Mostrar resultados
    print(f"Sugerencias únicas generadas: {sorted(list(sugerencias))}")
    print(f"Total de números sugeridos: {len(sugerencias)}")
    if detalles:
        res_df = pd.DataFrame(detalles).drop_duplicates(subset=['Numero']).sort_values(by='Sinergia_Hist', ascending=False)
        display(res_df[['Numero', 'Sinergia_Hist']])

generar_sugerencias_sinergicas(15)
```

```python
import numpy as np

def ranking_profesional_v2(target_type, dias_recientes=180):
    df_tipo = origins[origins['target_digit_type'] == target_type].copy()
    df_tipo['Fecha'] = pd.to_datetime(df_tipo['target_date'])

    # Identificador de regla
    df_tipo['regla'] = df_tipo['source_estado'] + ' ' + df_tipo['source_sesion'] + ' (' + df_tipo['source_number_type'] + '_' + df_tipo['source_digit_type'] + ')'

    # 1. Fuerza Histórica Estructural (Base sólida)
    scoring = df_tipo.groupby('regla')['target_date'].nunique().to_frame(name='Exitos_Totales')

    # 2. Fuerza de Momentum (Racha actual)
    max_fecha = df_tipo['Fecha'].max()
    fecha_corte = max_fecha - pd.Timedelta(days=dias_recientes)
    scoring['Exitos_Recientes'] = df_tipo[df_tipo['Fecha'] >= fecha_corte].groupby('regla')['target_date'].nunique()

    scoring = scoring.fillna(0)

    # Volvemos a la fórmula de veteranía: Peso mayor a la consistencia reciente sin ignorar la historia
    # Eliminamos cualquier sesgo de día de la semana no comprobado.
    scoring['Strength_Score'] = (scoring['Exitos_Recientes'] * 0.7) + (scoring['Exitos_Totales'] * 0.3)

    return scoring.sort_values('Strength_Score', ascending=False)

score_decena = ranking_profesional_v2('decena')
score_unidad = ranking_profesional_v2('unidad')

print("Ranking restaurado a lógica de Momentum e Historia (sin filtros de día).")
display(score_decena.head(5))
```

### El Modelo de 'Votación por Fuerza'
En lugar de generar un número por cada pareja, sumaremos la 'Fuerza' de todas las reglas que coincidan en un mismo número para hoy. Los números con mayor puntaje acumulado serán nuestro Top 25.

```python
def modelo_votacion_avanzado(n_reglas_considerar=25):
    # Obtener la fecha más reciente de los datos
    ultima_fecha = df_orig['Fecha'].max()
    datos_ayer = df_orig[df_orig['Fecha'] == ultima_fecha]

    # Tomar las reglas con mejor Strength_Score (Momento + Historia)
    top_decs = score_decena.head(n_reglas_considerar)
    top_unis = score_unidad.head(n_reglas_considerar)

    votos_numeros = {} # Numero -> Score Acumulado

    def get_val(regla_str, df_f):
        try:
            partes = regla_str.split(' ')
            e, s = partes[0], partes[1]
            contenido = regla_str.split('(')[1].split(')')[0]
            subpartes = contenido.split('_')
            dig_t = subpartes[-1]
            num_t = '_'.join(subpartes[:-1])

            match = df_f[(df_f['Estado'] == e) & (df_f['Sesión'] == s)]
            if not match.empty:
                val = match.iloc[0][num_t]
                return val[0] if dig_t == 'decena' else val[1]
        except:
            return None
        return None

    # Generar todos los cruces posibles y acumular su fuerza
    for r_d, row_d in top_decs.iterrows():
        val_d = get_val(r_d, datos_ayer)
        if val_d is None: continue

        for r_u, row_u in top_unis.iterrows():
            val_u = get_val(r_u, datos_ayer)
            if val_u is None: continue

            num_final = val_d + val_u
            # La fuerza del número es la suma de las fuerzas de las reglas que lo proponen
            fuerza_combinada = (row_d['Strength_Score'] + row_u['Strength_Score'])
            votos_numeros[num_final] = votos_numeros.get(num_final, 0) + fuerza_combinada

    # Crear ranking
    ranking_final = pd.DataFrame.from_dict(votos_numeros, orient='index', columns=['Puntaje_Fuerza'])
    ranking_final.index.name = 'Numero'
    ranking_final = ranking_final.sort_values('Puntaje_Fuerza', ascending=False).reset_index()

    print(f"Análisis para GA Midday basado en datos del {ultima_fecha.strftime('%Y-%m-%d')}")
    print(f"--- TOP 25 NÚMEROS CON MAYOR 'FUERZA CUÁNTICA' ---")
    display(ranking_final.head(25))
    return ranking_final

ranking_hoy = modelo_votacion_avanzado(25)
```

```python
import pandas as pd
from datetime import timedelta

def ejecutar_backtesting_profesional(dias_a_testear=30, n_reglas=25):
    fechas_test = ga_midday['Fecha'].sort_values().tail(dias_a_testear).tolist()
    exitos = 0
    total_procesado = 0

    print(f"Backtesting: Evaluando modelo de Momentum + Historia (Top {n_reglas})...\n")

    for fecha in fechas_test:
        ganador_real = ga_midday[ga_midday['Fecha'] == fecha]['Fijo'].values[0]
        fecha_ayer = fecha - timedelta(days=1)
        datos_ayer = df_orig[df_orig['Fecha'] == fecha_ayer]

        if datos_ayer.empty:
            continue

        votos_dia = {}

        def get_val_local(regla_str, df_f):
            try:
                partes = regla_str.split(' ')
                e, s = partes[0], partes[1]
                subpartes = regla_str.split('(')[1].split(')')[0].split('_')
                num_t = subpartes[0] + '_' + subpartes[1] if len(subpartes) == 3 else subpartes[0]
                dig_t = subpartes[-1]
                match_f = df_f[(df_f['Estado'] == e) & (df_f['Sesión'] == s)]
                if not match_f.empty:
                    val = match_f.iloc[0][num_t]
                    return val[0] if dig_t == 'decena' else val[1]
            except: return None
            return None

        # Ranking basado en la lógica restaurada
        for r_d, row_d in score_decena.head(n_reglas).iterrows():
            v_d = get_val_local(r_d, datos_ayer)
            if v_d is None: continue
            for r_u, row_u in score_unidad.head(n_reglas).iterrows():
                v_u = get_val_local(r_u, datos_ayer)
                if v_u is None: continue

                num = v_d + v_u
                fuerza = row_d['Strength_Score'] + row_u['Strength_Score']
                votos_dia[num] = votos_dia.get(num, 0) + fuerza

        top_25_dia = sorted(votos_dia, key=votos_dia.get, reverse=True)[:25]

        if ganador_real in top_25_dia:
            exitos += 1
            res = "✅"
        else:
            res = "❌"

        total_procesado += 1
        print(f"{fecha.strftime('%Y-%m-%d')} | G: {ganador_real} | {res}")

    precision = (exitos / total_procesado) * 100
    print(f"\nRESULTADO: {exitos}/{total_procesado} aciertos ({precision:.2f}%)")

ejecutar_backtesting_profesional(30)
```

```python
# 1. Analizar combinaciones específicas de reglas (Parejas de Éxito)
# Buscamos qué regla de Decena + qué regla de Unidad suelen acertar JUNTAS

def analizar_parejas_exito(n_top=10):
    # Pivotar para tener qué reglas acertaron cada día
    df_dec = origins[origins['target_digit_type'] == 'decena'].copy()
    df_dec['regla_dec'] = df_dec['source_estado'] + ' ' + df_dec['source_sesion'] + ' (' + df_dec['source_number_type'] + '_' + df_dec['source_digit_type'] + ')'

    df_uni = origins[origins['target_digit_type'] == 'unidad'].copy()
    df_uni['regla_uni'] = df_uni['source_estado'] + ' ' + df_uni['source_sesion'] + ' (' + df_uni['source_number_type'] + '_' + df_uni['source_digit_type'] + ')'

    # Cruzar por fecha
    merged = pd.merge(df_dec[['target_date', 'regla_dec']], df_uni[['target_date', 'regla_uni']], on='target_date')

    # Contar frecuencia de la pareja de reglas
    parejas = merged.groupby(['regla_dec', 'regla_uni']).size().reset_index(name='Frecuencia')
    parejas = parejas.sort_values(by='Frecuencia', ascending=False)

    print(f"--- TOP {n_top} PAREJAS DE REGLAS MÁS SINÉRGICAS ---")
    print("Estas combinaciones de origen producen el número exacto con mayor frecuencia:")
    display(parejas.head(n_top))
    return parejas

parejas_top = analizar_parejas_exito(15)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

# 1. Preparación de Features (X) y Target (y)
# Necesitamos convertir cada sorteo de GA Midday en una fila que contenga los resultados de todos los estados del día anterior

def preparar_dataset_ml(df_base, target_estado='GA', target_sesion='Midday'):
    data_ml = []
    # Obtener fechas únicas de los sorteos objetivo
    fechas = sorted(df_base[df_base['Estado'] == target_estado]['Fecha'].unique())

    for i in range(1, len(fechas)):
        hoy = fechas[i]
        ayer = hoy - pd.Timedelta(days=1)

        draw_hoy = df_base[(df_base['Fecha'] == hoy) & (df_base['Estado'] == target_estado) & (df_base['Sesión'] == target_sesion)]
        draws_ayer = df_base[df_base['Fecha'] == ayer]

        if draw_hoy.empty or draws_ayer.empty: continue

        # Construir el diccionario de la fila
        fila = {'target_fijo': draw_hoy.iloc[0]['Fijo'],
                'target_decena': int(draw_hoy.iloc[0]['Fijo'][0]),
                'target_unidad': int(draw_hoy.iloc[0]['Fijo'][1])}

        # Añadir todos los números de ayer como features
        for _, r in draws_ayer.iterrows():
            prefix = f"{r['Estado']}_{r['Sesión']}"
            fila[f"{prefix}_Fijo_D"] = int(r['Fijo'][0])
            fila[f"{prefix}_Fijo_U"] = int(r['Fijo'][1])
            fila[f"{prefix}_C1_D"] = int(r['Corrido_1'][0])
            fila[f"{prefix}_C1_U"] = int(r['Corrido_1'][1])
            fila[f"{prefix}_C2_D"] = int(r['Corrido_2'][0])
            fila[f"{prefix}_C2_U"] = int(r['Corrido_2'][1])

        data_ml.append(fila)

    return pd.DataFrame(data_ml).fillna(-1) # -1 si falta algún sorteo

df_ml = preparar_dataset_ml(df_orig)
print(f"Dataset para ML listo con {len(df_ml)} ejemplos.")
display(df_ml.head())
```

```python
# 2. Entrenamiento del Enjambre de Modelos
especialistas = {}
targets = ['target_dec_par', 'target_uni_par', 'target_dec_alto', 'target_uni_alto', 'target_suma_par']

for t in targets:
    # Usamos Gradient Boosting para capturar patrones no lineales complejos
    model = GradientBoostingClassifier(n_estimators=150, learning_rate=0.05, max_depth=5, random_state=42)
    model.fit(X_deep, df_deep[t])
    especialistas[t] = model
    print(f"Modelo especialista en {t} entrenado.")

# 3. Metamodelo de Combinación (Votación Ponderada)
def metamodelo_prediccion_avanzada():
    ultima_fecha = df_orig['Fecha'].max()
    draws_ayer = df_orig[df_orig['Fecha'] == ultima_fecha]

    # Preparar input
    input_dict = {}
    for _, r in draws_ayer.iterrows():
        prefix = f"{r['Estado']}_{r['Sesión']}"
        input_dict[f"{prefix}_Fijo_D"] = int(r['Fijo'][0])
        input_dict[f"{prefix}_Fijo_U"] = int(r['Fijo'][1])
        input_dict[f"{prefix}_C1_D"] = int(r['Corrido_1'][0])
        input_dict[f"{prefix}_C1_U"] = int(r['Corrido_1'][1])
        input_dict[f"{prefix}_C2_D"] = int(r['Corrido_2'][0])
        input_dict[f"{prefix}_C2_U"] = int(r['Corrido_2'][1])

    X_input = pd.DataFrame([input_dict]).reindex(columns=X_deep.columns, fill_value=-1)

    # Recolectar 'votos' de los especialistas
    scores_finales = np.zeros(100)

    # Probabilidades de cada dimensión
    p_dec_par = especialistas['target_dec_par'].predict_proba(X_input)[0][1]
    p_uni_par = especialistas['target_uni_par'].predict_proba(X_input)[0][1]
    p_dec_alt = especialistas['target_dec_alto'].predict_proba(X_input)[0][1]
    p_uni_alt = especialistas['target_uni_alto'].predict_proba(X_input)[0][1]
    p_sum_par = especialistas['target_suma_par'].predict_proba(X_input)[0][1]

    for n in range(100):
        d, u = divmod(n, 10)
        # Calcular probabilidad compuesta del número n
        score = 1.0
        score *= (p_dec_par if d % 2 != 0 else 1 - p_dec_par)
        score *= (p_uni_par if u % 2 != 0 else 1 - p_uni_par)
        score *= (p_dec_alt if d >= 5 else 1 - p_dec_alt)
        score *= (p_uni_alt if u >= 5 else 1 - p_uni_alt)
        score *= (p_sum_par if (d + u) % 2 != 0 else 1 - p_sum_par)

        scores_finales[n] = score

    # Obtener Top 25
    top_indices = scores_finales.argsort()[-25:][::-1]
    predicciones = [f"{i:02d}" for i in top_indices]

    print(f"\n--- PREDICCIÓN METAMODELO DEEP (BASADA EN EL CAOS DEL {ultima_fecha.date()}) ---")
    print(f"TOP 25 SUGERIDO: {predicciones}")
    return predicciones

top_25_deep = metamodelo_prediccion_avanzada()
```

```python
def metamodelo_hibrido_con_heuristicas(prob_ensemble, top_parejas, datos_ayer):
    # prob_ensemble: array de 100 con scores del Deep Ensemble
    # top_parejas: dataframe con las parejas más sinérgicas

    scores_hibridos = prob_ensemble.copy()

    # Extraer números sugeridos por las reglas de sinergia para hoy
    votos_reglas = {}
    for _, row in top_parejas.head(20).iterrows():
        # Función de ayuda para obtener el dígito de la regla
        def get_d(r_str, df_f):
            try:
                p = r_str.split(' ')
                e, s = p[0], p[1]
                sub = r_str.split('(')[1].split(')')[0].split('_')
                num_t = sub[0] + '_' + sub[1] if len(sub) == 3 else sub[0]
                dig_t = sub[-1]
                match = df_f[(df_f['Estado'] == e) & (df_f['Sesión'] == s)]
                if not match.empty:
                    val = match.iloc[0][num_t]
                    return int(val[0] if dig_t == 'decena' else val[1])
            except: return None
            return None

        d = get_d(row['regla_dec'], datos_ayer)
        u = get_d(row['regla_uni'], datos_ayer)

        if d is not null and u is not null:
            n = d * 10 + u
            # Aumentamos el score del ensemble basándonos en la frecuencia histórica de la regla
            votos_reglas[n] = votos_reglas.get(n, 0) + (row['Frecuencia'] / 100.0)

    # Combinar: El score final es el del ensemble + el bono de las reglas
    for n, bono in votos_reglas.items():
        scores_hibridos[n] += bono

    return scores_hibridos

print("Capa de Heurísticas de Sinergia integrada al Metamodelo.")
```

### Objetivo: +70% Precisión en Top 25 GA Midday
Este sistema utiliza una arquitectura de 'Especialistas Dinámicos' que desglosan el mundo de 100 números en dimensiones binarias antes de la síntesis final.

```python
import pandas as pd
import numpy as np
import xgboost as xgb

# 1. Asegurar Carga de Datos Robusta
file_path = "US_Lottery_Draws_Master_modified.csv"
df_master = pd.read_csv(file_path, dtype=str)
df_master.columns = df_master.columns.str.strip()
df_master['Fecha'] = pd.to_datetime(df_master['Fecha'], errors='coerce')
for col in ['Fijo', 'Corrido_1', 'Corrido_2']:
    df_master[col] = df_master[col].astype(str).str.strip().str.zfill(2)
df_master = df_master.dropna(subset=['Fecha', 'Fijo'])

# 2. Construcción de Features con Swarm Intelligence (Sin Intuición)
def build_swarm_dataset(df_base):
    df_ga = df_base[(df_base['Estado'] == 'GA') & (df_base['Sesión'] == 'Midday')].sort_values('Fecha')
    processed = []

    for i in range(5, len(df_ga)):
        hoy = df_ga.iloc[i]
        try:
            val_fijo = int(hoy['Fijo'])
            ayer_f = hoy['Fecha'] - pd.Timedelta(days=1)
            contexto_ayer = df_base[df_base['Fecha'] == ayer_f]
            if contexto_ayer.empty: continue

            # Especialistas Binarios (Heurística de Jugador)
            features = {
                'target': val_fijo,
                't_paridad': val_fijo % 2,
                't_alto': 1 if val_fijo >= 50 else 0,
                't_suma_par': (int(hoy['Fijo'][0]) + int(hoy['Fijo'][1])) % 2,
                't_decena': int(hoy['Fijo'][0]),
                't_unidad': int(hoy['Fijo'][1])
            }

            # Inyección de Contexto Total de Ayer
            for _, r in contexto_ayer.iterrows():
                p = f"{r['Estado']}_{r['Sesión']}"
                features[f"{p}_f"] = int(r['Fijo'])
                features[f"{p}_c1"] = int(r['Corrido_1'])
                features[f"{p}_c2"] = int(r['Corrido_2'])
            processed.append(features)
        except: continue
    return pd.DataFrame(processed).fillna(-1)

print("Construyendo Dataset de Alta Dimensión para Swarm Intelligence...")
df_swarm = build_swarm_dataset(df_master)
display(df_swarm.tail(2))
```

```python
def entrenar_comite_avanzado(df_in):
    target_cols = [c for c in df_in.columns if c.startswith('t_') or c == 'target']
    X = df_in.drop(target_cols, axis=1)

    # El hoy manda: Pesos exponenciales para el contexto reciente
    weights = np.exp(np.linspace(-4, 0, len(df_in)))

    comite = {}
    especialidades = ['t_paridad', 't_alto', 't_suma_par', 't_decena', 't_unidad']

    print("Entrenando Comité de Especialistas XGBoost Swarm...")
    for t in especialidades:
        is_multi = 'decena' in t or 'unidad' in t
        clf = xgb.XGBClassifier(
            n_estimators=400,
            max_depth=9,
            learning_rate=0.015,
            objective='multi:softprob' if is_multi else 'binary:logistic',
            subsample=0.8,
            colsample_bytree=0.8,
            random_state=42,
            tree_method='hist'
        )
        clf.fit(X, df_in[t], sample_weight=weights)
        comite[t] = clf

    return comite, X.columns

comite_modelos, feature_cols_swarm = entrenar_comite_avanzado(df_swarm)
print("Comité Swarm de alta precisión listo.")
```

```python
def deliberacion_swarm(n_top=25):
    # 1. Obtener contexto de ayer
    ultima_fecha = df_master['Fecha'].max()
    ayer_data = df_master[df_master['Fecha'] == ultima_fecha]

    input_dict = {}
    for _, r in ayer_data.iterrows():
        prefix = f"{r['Estado']}_{r['Sesión']}"
        input_dict[f"{prefix}_f"] = int(r['Fijo'])
        input_dict[f"{prefix}_c1"] = int(r['Corrido_1'])
        input_dict[f"{prefix}_c2"] = int(r['Corrido_2'])

    # Alinear con las columnas del entrenamiento
    X_predict = pd.DataFrame([input_dict]).reindex(columns=feature_cols_swarm, fill_value=-1)

    # 2. Extraer probabilidades de los especialistas
    p_par = comite_modelos['t_paridad'].predict_proba(X_predict)[0][1]
    p_alt = comite_modelos['t_alto'].predict_proba(X_predict)[0][1]
    p_sum = comite_modelos['t_suma_par'].predict_proba(X_predict)[0][1]
    p_dec = comite_modelos['t_decena'].predict_proba(X_predict)[0]
    p_uni = comite_modelos['t_unidad'].predict_proba(X_predict)[0]

    # 3. Calcular Score de Consenso para cada número (00-99)
    scores = np.zeros(100)
    for n in range(100):
        d, u = divmod(n, 10)
        # Probabilidad base (Posicional)
        score = p_dec[d] * p_uni[u]

        # Multiplicadores de Filtros Binarios (Refuerzo de Heurística)
        score *= (p_par if n % 2 != 0 else (1 - p_par))
        score *= (p_alt if n >= 50 else (1 - p_alt))
        score *= (p_sum if (d + u) % 2 != 0 else (1 - p_sum))

        scores[n] = score

    # 4. Generar Ranking
    top_indices = scores.argsort()[-n_top:][::-1]
    sugeridos = [f"{i:02d}" for i in top_indices]

    print(f"--- RESULTADO SWARM INTELLIGENCE: GA MIDDAY ---")
    print(f"Basado en Contexto del: {ultima_fecha.date()}")
    print(f"Top {n_top} Recomendados: {sugeridos}")
    print(f"Confianza en el número líder ({sugeridos[0]}): {scores[top_indices[0]]*100:.2f}%")

deliberacion_swarm(25)
```

```python
def backtest_swarm_30_dias(n_dias=30):
    ga_midday_test = df_master[(df_master['Estado'] == 'GA') & (df_master['Sesión'] == 'Midday')].sort_values('Fecha')
    fechas_test = ga_midday_test['Fecha'].tail(n_dias).tolist()

    exitos = 0
    total = 0

    print(f'🚀 Iniciando Backtesting de Alta Precisión (Últimos {n_dias} días)...')

    for fecha in fechas_test:
        ganador_real = ga_midday_test[ga_midday_test['Fecha'] == fecha]['Fijo'].values[0]
        ayer_f = fecha - pd.Timedelta(days=1)
        contexto_ayer = df_master[df_master['Fecha'] == ayer_f]

        if contexto_ayer.empty: continue

        # Simular deliberación del Swarm
        input_dict = {}
        for _, r in contexto_ayer.iterrows():
            p = f"{r['Estado']}_{r['Sesión']}"
            input_dict[f"{p}_f"] = int(r['Fijo'])
            input_dict[f"{p}_c1"] = int(r['Corrido_1'])
            input_dict[f" {p}_c2"] = int(r['Corrido_2'])

        X_test = pd.DataFrame([input_dict]).reindex(columns=feature_cols_swarm, fill_value=-1)

        # Score de Consenso
        p_par = comite_modelos['t_paridad'].predict_proba(X_test)[0][1]
        p_alt = comite_modelos['t_alto'].predict_proba(X_test)[0][1]
        p_sum = comite_modelos['t_suma_par'].predict_proba(X_test)[0][1]
        p_dec = comite_modelos['t_decena'].predict_proba(X_test)[0]
        p_uni = comite_modelos['t_unidad'].predict_proba(X_test)[0]

        scores = np.zeros(100)
        for n in range(100):
            d, u = divmod(n, 10)
            s = p_dec[d] * p_uni[u]
            s *= (p_par if n % 2 != 0 else (1 - p_par))
            s *= (p_alt if n >= 50 else (1 - p_alt))
            s *= (p_sum if (d + u) % 2 != 0 else (1 - p_sum))
            scores[n] = s

        top_25 = [f"{i:02d}" for i in scores.argsort()[-25:]]

        if ganador_real in top_25:
            exitos += 1
        total += 1

    print(f'\n--- REPORTE DE RENDIMIENTO SWARM ---')
    print(f'Aciertos en Top 25: {exitos}/{total}')
    print(f'Precisión Verificada: {(exitos/total)*100:.2f}%')

backtest_swarm_30_dias(30)
```

```python
def backtest_blindado_top_15(n_dias=20):
    # Este test elimina el 'Data Leakage' re-entrenando el modelo en cada paso
    ga_test = df_master[(df_master['Estado'] == 'GA') & (df_master['Sesión'] == 'Midday')].sort_values('Fecha')
    fechas_eval = ga_test['Fecha'].tail(n_dias).tolist()

    aciertos_top15 = 0
    aciertos_top25 = 0
    total = 0

    print(f'🛡️ Iniciando Validación Ciega (Sin Fuga de Datos) - Top 15 vs Top 25...')

    for i, fecha_objetivo in enumerate(fechas_eval):
        # 1. Definir set de entrenamiento: SOLO datos ANTERIORES a la fecha objetivo
        df_entrenamiento = df_swarm[pd.to_datetime(df_master.loc[df_swarm.index, 'Fecha']) < fecha_objetivo]

        if len(df_entrenamiento) < 100: continue

        # 2. Re-entrenar especialistas de forma rápida
        X_train = df_entrenamiento.drop([c for c in df_entrenamiento.columns if c.startswith('t_') or c == 'target'], axis=1)
        w = np.exp(np.linspace(-3, 0, len(df_entrenamiento)))

        comite_wf = {}
        for t in ['t_paridad', 't_alto', 't_suma_par', 't_decena', 't_unidad']:
            is_multi = 'decena' in t or 'unidad' in t
            m = xgb.XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.05,
                                  objective='multi:softprob' if is_multi else 'binary:logistic',
                                  random_state=42, tree_method='hist')
            m.fit(X_train, df_entrenamiento[t], sample_weight=w)
            comite_wf[t] = m

        # 3. Predecir usando datos de 'ayer'
        ayer_f = fecha_objetivo - pd.Timedelta(days=1)
        datos_ayer = df_master[df_master['Fecha'] == ayer_f]
        ganador_real = ga_test[ga_test['Fecha'] == fecha_objetivo]['Fijo'].values[0]

        input_dict = {}
        for _, r in datos_ayer.iterrows():
            p = f"{r['Estado']}_{r['Sesión']}"
            input_dict[f"{p}_f"] = int(r['Fijo'])
            input_dict[f"{p}_c1"] = int(r['Corrido_1'])
            input_dict[f"{p}_c2"] = int(r['Corrido_2'])

        X_p = pd.DataFrame([input_dict]).reindex(columns=X_train.columns, fill_value=-1)

        # 4. Score de Consenso
        p_p, p_a, p_s = comite_wf['t_paridad'].predict_proba(X_p)[0][1], comite_wf['t_alto'].predict_proba(X_p)[0][1], comite_wf['t_suma_par'].predict_proba(X_p)[0][1]
        p_d, p_u = comite_wf['t_decena'].predict_proba(X_p)[0], comite_wf['t_unidad'].predict_proba(X_p)[0]

        scores = np.zeros(100)
        for n in range(100):
            d, u = divmod(n, 10)
            s = p_d[d] * p_u[u]
            s *= (p_p if n % 2 != 0 else (1 - p_p))
            s *= (p_a if n >= 50 else (1 - p_a))
            s *= (p_s if (d + u) % 2 != 0 else (1 - p_s))
            scores[n] = s

        top_25 = [f"{i:02d}" for i in scores.argsort()[-25:]]
        top_15 = top_25[-15:]

        if ganador_real in top_25: aciertos_top25 += 1
        if ganador_real in top_15: aciertos_top15 += 1
        total += 1
        print(f".", end="")

    print(f'\n\n--- RESULTADOS FINALES DE ALTA FIDELIDAD ---')
    print(f'Precisión Real Top 25: {(aciertos_top25/total)*100:.2f}%')
    print(f'Precisión Real Top 15: {(aciertos_top15/total)*100:.2f}%')

backtest_blindado_top_15(20)
```

```python
def auditoria_final_top10(n_dias=15):
    ga_test = df_master[(df_master['Estado'] == 'GA') & (df_master['Ses_n'] == 'Midday') if 'Ses_n' in df_master.columns else (df_master['Estado'] == 'GA') & (df_master['Sesión'] == 'Midday')].sort_values('Fecha')
    fechas_eval = ga_test['Fecha'].tail(n_dias).tolist()

    aciertos_top10 = 0
    total = 0

    print(f'🕵️ AUDITORÍA DE ESTRÉS: TOP 10 (Sin Fuga de Datos)')
    print(f'--------------------------------------------------')

    for fecha_obj in fechas_eval:
        # FILTRO CRÍTICO: El modelo es Ciego al futuro
        fecha_str = fecha_obj.strftime('%Y-%m-%d')
        df_past = df_swarm[pd.to_datetime(df_master.loc[df_swarm.index, 'Fecha']) < fecha_obj]

        # Re-entrenamiento flash
        X_t = df_past.drop([c for c in df_past.columns if c.startswith('t_') or c == 'target'], axis=1)
        comite_audit = {}
        for t in ['t_paridad', 't_alto', 't_suma_par', 't_decena', 't_unidad']:
            m = xgb.XGBClassifier(n_estimators=100, max_depth=5, learning_rate=0.1, random_state=42, tree_method='hist')
            m.fit(X_t, df_past[t])
            comite_audit[t] = m

        # Predicción
        ayer = fecha_obj - pd.Timedelta(days=1)
        d_ayer = df_master[df_master['Fecha'] == ayer]
        ganador = ga_test[ga_test['Fecha'] == fecha_obj]['Fijo'].values[0]

        in_dict = {}
        for _, r in d_ayer.iterrows():
            p = f"{r['Estado']}_{r['Sesión']}"
            in_dict[f"{p}_f"] = int(r['Fijo'])
            in_dict[f"{p}_c1"] = int(r['Corrido_1'])
            in_dict[f"{p}_c2"] = int(r['Corrido_2'])

        X_p = pd.DataFrame([in_dict]).reindex(columns=X_t.columns, fill_value=-1)

        # Fusión de Probabilidades
        p_p, p_a, p_s = comite_audit['t_paridad'].predict_proba(X_p)[0][1], comite_audit['t_alto'].predict_proba(X_p)[0][1], comite_audit['t_suma_par'].predict_proba(X_p)[0][1]
        p_d, p_u = comite_audit['t_decena'].predict_proba(X_p)[0], comite_audit['t_unidad'].predict_proba(X_p)[0]

        scores = np.zeros(100)
        for n in range(100):
            d, u = divmod(n, 10)
            s = p_d[d] * p_u[u]
            s *= (p_p if n % 2 != 0 else (1 - p_p))
            s *= (p_a if n >= 50 else (1 - p_a))
            s *= (p_s if (d + u) % 2 != 0 else (1 - p_s))
            scores[n] = s

        top_10 = [f"{i:02d}" for i in scores.argsort()[-10:]]
        res = "✅" if ganador in top_10 else "❌"
        if ganador in top_10: aciertos_top10 += 1
        total += 1
        print(f"Fecha: {fecha_str} | Ganador: {ganador} | En Top 10: {res}")

    print(f'--------------------------------------------------')
    print(f'RESULTADO AUDITORÍA TOP 10: {(aciertos_top10/total)*100:.2f}%')

auditoria_final_top10(15)
```

```python
def auditoria_magistral_500_dias(n_dias=500):
    ga_test = df_master[(df_master['Estado'] == 'GA') & (df_master['Sesión'] == 'Midday')].sort_values('Fecha')
    fechas_eval = ga_test['Fecha'].tail(n_dias).tolist()

    stats = {'top5': 0, 'top10': 0, 'top15': 0, 'top25': 0}
    total = 0

    print(f'🚀 INICIANDO AUDITORÍA MAGISTRAL: {n_dias} DÍAS (Walk-Forward Estricto)')
    print('Analizando integridad y precisión por niveles...')

    for i, fecha_obj in enumerate(fechas_eval):
        # 1. Aislamiento total: Solo datos estrictamente pasados
        df_past = df_swarm[pd.to_datetime(df_master.loc[df_swarm.index, 'Fecha']) < fecha_obj]

        if len(df_past) < 200: continue

        # 2. Re-entrenamiento del Comité
        X_t = df_past.drop([c for c in df_past.columns if c.startswith('t_') or c == 'target'], axis=1)
        w = np.exp(np.linspace(-2.3, 0, len(df_past)))

        comite_wf = {}
        for t in ['t_paridad', 't_alto', 't_suma_par', 't_decena', 't_unidad']:
            is_multi = 'decena' in t or 'unidad' in t
            m = xgb.XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.1,
                                  objective='multi:softprob' if is_multi else 'binary:logistic',
                                  random_state=42, tree_method='hist')
            m.fit(X_t, df_past[t], sample_weight=w)
            comite_wf[t] = m

        # 3. Predicción con datos de 'Ayer'
        ayer = fecha_obj - pd.Timedelta(days=1)
        d_ayer = df_master[df_master['Fecha'] == ayer]
        ganador = ga_test[ga_test['Fecha'] == fecha_obj]['Fijo'].values[0]

        in_dict = {}
        for _, r in d_ayer.iterrows():
            p = f"{r['Estado']}_{r['Sesión']}"
            # Manejo robusto de NaN convirtiendo a -1 si el dato no existe
            try:
                in_dict[f"{p}_f"] = int(r['Fijo']) if pd.notnull(r['Fijo']) and str(r['Fijo']).lower() != 'nan' else -1
                in_dict[f"{p}_c1"] = int(r['Corrido_1']) if pd.notnull(r['Corrido_1']) and str(r['Corrido_1']).lower() != 'nan' else -1
                in_dict[f"{p}_c2"] = int(r['Corrido_2']) if pd.notnull(r['Corrido_2']) and str(r['Corrido_2']).lower() != 'nan' else -1
            except:
                continue

        X_p = pd.DataFrame([in_dict]).reindex(columns=X_t.columns, fill_value=-1)

        # 4. Fusión Swarm
        p_p, p_a, p_s = comite_wf['t_paridad'].predict_proba(X_p)[0][1], comite_wf['t_alto'].predict_proba(X_p)[0][1], comite_wf['t_suma_par'].predict_proba(X_p)[0][1]
        p_d, p_u = comite_wf['t_decena'].predict_proba(X_p)[0], comite_wf['t_unidad'].predict_proba(X_p)[0]

        scores = np.zeros(100)
        for n in range(100):
            d, u = divmod(n, 10)
            s = p_d[d] * p_u[u]
            s *= (p_p if n % 2 != 0 else (1 - p_p))
            s *= (p_a if n >= 50 else (1 - p_a))
            s *= (p_s if (d + u) % 2 != 0 else (1 - p_s))
            scores[n] = s

        ranking = [f"{idx:02d}" for idx in scores.argsort()[::-1]]

        if ganador in ranking[:5]: stats['top5'] += 1
        if ganador in ranking[:10]: stats['top10'] += 1
        if ganador in ranking[:15]: stats['top15'] += 1
        if ganador in ranking[:25]: stats['top25'] += 1
        total += 1

        if i % 50 == 0: print(f"Progreso: {i}/{n_dias} días procesados...")

    print(f'\n--- REPORTE FINAL DE ESTRÉS ({total} DÍAS VÁLIDOS) ---')
    for k, v in stats.items():
        print(f"Precisión en {k.upper()}: {(v/total)*100:.2f}% ({v}/{total})")

auditoria_magistral_500_dias(500)
```
